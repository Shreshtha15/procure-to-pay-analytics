# Basic Dataset Profiling

```python
# 1. Count unique purchase-item cases
print("Unique purchase-item cases:",
      log["case:concept:name"].nunique())
```

```python
# 2. Count unique purchase documents
print("Unique purchase documents:",
      log["case:Purchasing Document"].nunique())
```

```python
# 3. Count unique vendors
print("Unique vendors:",
      log["case:Vendor"].nunique())
```

```python
# 4. Count unique event activities
print("Unique activities:",
      log["concept:name"].nunique())
```

```python
# 5. Inspect the activities and their frequencies
print("\nMost frequent activities:")
print(log["concept:name"].value_counts().head(15))
```

```python
# 6. Inspect the timestamp range
print("\nEarliest event:", log["time:timestamp"].min())
print("Latest event:", log["time:timestamp"].max())
```

```python
# 7. Check for completely duplicated rows
print("\nCompletely duplicated rows:",
      log.duplicated().sum())
```
